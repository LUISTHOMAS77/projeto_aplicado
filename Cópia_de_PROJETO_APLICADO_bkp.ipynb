```python
import pandas as pd
import glob

# Caminho dos arquivos CSV
caminho_arquivos = '/content/drive/MyDrive/dados/*.csv'
lista_arquivos = glob.glob(caminho_arquivos)

dataframes_filtrados = []

# Definição do filtro
coluna_para_filtrar = 'CD_OPERADORA'
valor_do_filtro = '352501'

for arquivo in lista_arquivos:
    print(f"Processando: {arquivo}")

    # Define o separador como ponto e vírgula (sep=';') e a codificação (encoding='latin1')
    df_temp = pd.read_csv(arquivo, sep=';', encoding='latin1', low_memory=False)

    # Converte a coluna para texto para evitar incompatibilidade de tipos
    df_temp[coluna_para_filtrar] = df_temp[coluna_para_filtrar].astype(str)

    # Aplica o filtro
    df_temp_filtrado = df_temp[df_temp[coluna_para_filtrar] == valor_do_filtro]

    dataframes_filtrados.append(df_temp_filtrado)

# Junta todos os dataframes filtrados em um único
df_final = pd.concat(dataframes_filtrados, ignore_index=True)
print(f"Total de linhas após o filtro: {len(df_final)}")
```

```python
import pandas as pd
import numpy as np
import re # Necessário para a função converter_faixa_etaria

# --- Início: Garantir a presença de 'idade_ponto_medio' ---
# Esta função foi copiada de uma célula anterior para garantir que a coluna exista
def converter_faixa_etaria(valor):
  if pd.isna(valor) or 'incorr' in str(valor).lower():
    return np.nan
  texto = str(valor).strip().lower()
  if 'menos de 1' in texto:
    return 0.5
  if '80' in texto and 'mais' in texto:
    return 82.5
  numeros = re.findall(r'\d+', texto)
  if len(numeros) == 2:
    inicio, fim = float(numeros[0]), float(numeros[1])
    return (inicio + fim) / 2.0
  return np.nan

# Aplica a função para criar 'idade_ponto_medio' se ela não existir no DataFrame
# Isso é crucial para que o pd.cut funcione.
if 'idade_ponto_medio' not in df_perfil.columns:
    df_perfil['idade_ponto_medio'] = df_perfil['DE_FAIXA_ETARIA'].apply(converter_faixa_etaria)
# --- Fim: Garantir a presença de 'idade_ponto_medio' ---


# Define the bins and labels for the consolidated age groups
bins = [0, 18, 38, 58, np.inf]
labels = ['0-18', '19-38', '39-58', '59+']

# Create the new consolidated age group column using 'idade_ponto_medio'
# The 'right=True' means the rightmost bin edge is included (e.g., <=18).
# 'include_lowest=True' ensures that the lowest value is included in the first bin.
df_perfil['DE_FAIXA_ETARIA_CONSOLIDADA'] = pd.cut(
    df_perfil['idade_ponto_medio'],
    bins=bins,
    labels=labels,
    right=True,
    include_lowest=True,
    ordered=True # Keep the order of categories
)

# Display the value counts of the new column to verify
print("Contagem das novas faixas etárias consolidadas:")
print(df_perfil['DE_FAIXA_ETARIA_CONSOLIDADA'].value_counts().sort_index())

# Drop the original 'DE_FAIXA_ETARIA' and 'idade_ponto_medio' columns as requested by "seja substituida"
df_perfil.drop(columns=['DE_FAIXA_ETARIA', 'idade_ponto_medio'], inplace=True, errors='ignore')

print("\nDataFrame df_perfil após a consolidação das faixas etárias:")
print(df_perfil[['DE_FAIXA_ETARIA_CONSOLIDADA']].head())
print(f"Shape atual do DataFrame: {df_perfil.shape}")
```

```python
import pandas as pd
import numpy as np

# --- Step 1: Dummify 'DE_FAIXA_ETARIA_CONSOLIDADA' if it exists as a categorical/object type ---
# The kernel state shows 'DE_FAIXA_ETARIA_CONSOLIDADA' is present in df_perfil
# and previous steps (GjSBswqgcmib) created it as a 'category' or 'object'.
# Check if it exists and is indeed a categorical type before dummifying.
if 'DE_FAIXA_ETARIA_CONSOLIDADA' in df_perfil.columns and (df_perfil['DE_FAIXA_ETARIA_CONSOLIDADA'].dtype == 'category' or df_perfil['DE_FAIXA_ETARIA_CONSOLIDADA'].dtype == 'object'):
    print("Dummifying 'DE_FAIXA_ETARIA_CONSOLIDADA'...")
    df_perfil = pd.get_dummies(
        df_perfil,
        columns=['DE_FAIXA_ETARIA_CONSOLIDADA'],
        prefix='DE_FAIXA_ETARIA_CONSOLIDADA',
        drop_first=False, # Keep all categories
        dtype=int
    )
    print("Dummy variables created for DE_FAIXA_ETARIA_CONSOLIDADA.")
# Add a check to see if it's already dummified (e.g., one of its dummy columns exists)
elif any(col.startswith('DE_FAIXA_ETARIA_CONSOLIDADA_') for col in df_perfil.columns):
    print("DE_FAIXA_ETARIA_CONSOLIDADA appears to be already dummified (dummy columns exist).")
else:
    print("Warning: 'DE_FAIXA_ETARIA_CONSOLIDADA' not found or not in expected categorical format for dummification. Skipping dummification for this column.")


# --- Step 2: Define aggregation dictionary for all other columns ---
groupby_col = 'ID_CMPT_MOVEL'
agg_dict = {}

# Columns to be explicitly excluded from aggregation as they are either IDs,
# original categoricals now replaced by dummies, or temporary columns that should not be aggregated directly.
cols_to_exclude_from_agg = [
    'CD_MUNICIPIO', 'NM_MUNICIPIO', 'TP_SEXO', 'DE_FAIXA_ETARIA',
    'DE_FAIXA_ETARIA_REAJ', 'CD_PLANO', 'DT_CARGA', 'mes_ano_iso', 'Reclamações',
    # Explicitly exclude the original categorical columns that have been dummified:
    'TP_VIGENCIA_PLANO', 'DE_CONTRATACAO_PLANO', 'DE_SEGMENTACAO_PLANO',
    'DE_ABRG_GEOGRAFICA_PLANO', 'COBERTURA_ASSIST_PLAN', 'TIPO_VINCULO',
    'idade_ponto_medio' # 'idade_ponto_medio' is not in df_perfil based on kernel state.
]

for col in df_perfil.columns:
    if col == groupby_col:
        continue
    if col in cols_to_exclude_from_agg:
        continue

    # Aggregate quantitative beneficiary counts by sum
    if col in ['QT_BENEFICIARIO_ATIVO', 'QT_BENEFICIARIO_ADERIDO', 'QT_BENEFICIARIO_CANCELADO']:
        agg_dict[col] = 'sum'
    # Aggregate dummy/boolean columns (including newly dummified DE_FAIXA_ETARIA_CONSOLIDADA,
    # and existing NM_MUNICIPIO_..., TP_SEXO_M, DE_CONTRATACAO_PLANO_...) by both sum (count) and mean (proportion).
    # We explicitly look for columns that are 0/1 integers or bools.
    elif pd.api.types.is_bool_dtype(df_perfil[col]) or \
         (pd.api.types.is_integer_dtype(df_perfil[col]) and df_perfil[col].isin([0, 1]).all()):
        agg_dict[col] = ['sum', 'mean']
    # For any other numerical columns that haven't been explicitly handled, aggregate by mean.
    elif pd.api.types.is_numeric_dtype(df_perfil[col]):
        agg_dict[col] = 'mean'
    else:
        print(f"Warning: Column '{col}' of type '{df_perfil[col].dtype}' is not suitable for aggregation and will be excluded.")

# --- Step 3: Perform groupby aggregation ---
if agg_dict:
    df_agrupado = df_perfil.groupby(groupby_col).agg(agg_dict)

    # Flatten multi-level columns due to ['sum', 'mean'] aggregation
    df_agrupado.columns = ['_'.join(col).strip() if isinstance(col, tuple) else col for col in df_agrupado.columns.values]

    # Display results
    print(f"\nDataFrame agrupado por {groupby_col}:")
    print(df_agrupado.head())
    print(f"Shape do DataFrame agrupado: {df_agrupado.shape}")
    print("\nColunas do DataFrame agrupado:")
    print(df_agrupado.columns.tolist())
else:
    print("No suitable columns found for aggregation. Please check the DataFrame content and `cols_to_exclude_from_agg`.")
```

```python
import statsmodels.api as sm

def stepwise_backward_pvalue(df, target, p_limite=0.05):
    y = df[target]
    X = df.drop(columns=[target])
    X_const = sm.add_constant(X)

    modelo = sm.OLS(y, X_const).fit()
    pvalues = modelo.pvalues.drop('const')

    while pvalues.max() > p_limite:
        var_remover = pvalues.idxmax()
        print(f"Removendo variável: {var_remover} (p = {pvalues.max():.4f})")

        X = X.drop(columns=[var_remover])
        X_const = sm.add_constant(X)
        modelo = sm.OLS(y, X_const).fit()
        pvalues = modelo.pvalues.drop('const')

    return modelo, X.columns
```

```python
import pandas as pd
import numpy as np

# 'idade_ponto_medio' is already in df_perfil from previous steps.
# 'TP_SEXO_M' is already in df_perfil from previous steps.
# 'NM_MUNICIPIO_agrupado_...' columns are already in df_perfil from previous steps.

# 1. Create grouped age categories from 'idade_ponto_medio' and dummify
bins = [0, 18, 38, 58, np.inf]
labels = ['0-18', '19-38', '39-58', '59+']
df_perfil['faixa_etaria_grupo'] = pd.cut(
    df_perfil['idade_ponto_medio'], bins=bins, labels=labels, right=True, include_lowest=True
)

# Convert age groups to dummy variables and add to the DataFrame
df_perfil = pd.get_dummies(
    df_perfil, columns=['faixa_etaria_grupo'], prefix='idade_grupo', drop_first=False, dtype=int
)

# 2. Dummify specified categorical plan columns and add to the DataFrame
colunas_planos = [
    'DE_CONTRATACAO_PLANO',
    'DE_SEGMENTACAO_PLANO',
    'DE_ABRG_GEOGRAFICA_PLANO',
    'COBERTURA_ASSIST_PLAN',
    'TIPO_VINCULO',
    'TP_VIGENCIA_PLANO' # Also a categorical column from the dataset
]

# Filter for columns that actually exist in the DataFrame before dummifying
existing_plan_cols_to_dummy = [col for col in colunas_planos if col in df_perfil.columns]

df_perfil = pd.get_dummies(
    df_perfil, columns=existing_plan_cols_to_dummy, prefix='plano', drop_first=False, dtype=int
)

# Define aggregation dictionary
# Group by 'ID_CMPT_MOVEL'

# Identify all dummy columns created (age groups, plan types, and existing municipality dummies)
dummy_cols_for_agg = [
    col for col in df_perfil.columns
    if col.startswith(('idade_grupo_', 'plano_', 'NM_MUNICIPIO_agrupado_'))
]

agg_dict = {
    'Reclamações': 'sum', # Total de reclamações
    'idade_ponto_medio': 'mean', # Média de idade
    'TP_SEXO_M': ['sum', 'mean'], # Contagem de homens e proporção de homens
    'QT_BENEFICIARIO_ATIVO': 'sum',
    'QT_BENEFICIARIO_ADERIDO': 'sum',
    'QT_BENEFICIARIO_CANCELADO': 'sum',
}

# Add dummy columns to aggregation dictionary for sum (count) and mean (proportion)
for col in dummy_cols_for_agg:
    # Ensure the column exists before adding to agg_dict
    if col in df_perfil.columns:
        agg_dict[col] = ['sum', 'mean']

# Perform the groupby aggregation
df_agrupado = df_perfil.groupby('ID_CMPT_MOVEL').agg(agg_dict)

# Flatten multi-level columns due to ['sum', 'mean'] aggregation
df_agrupado.columns = ['_'.join(col).strip() for col in df_agrupado.columns.values]

# Calculate derived metrics for clarity and user request compliance
# Proportion of female
# Check if TP_SEXO_M_mean exists before creating TP_SEXO_F_proportion
if 'TP_SEXO_M_mean' in df_agrupado.columns:
    df_agrupado['TP_SEXO_F_proportion'] = 1 - df_agrupado['TP_SEXO_M_mean']

# Rename columns for clarity based on user's request for municipality and gender
df_agrupado.rename(columns={
    'TP_SEXO_M_mean': 'TP_SEXO_M_proportion',
    'TP_SEXO_M_sum': 'TP_SEXO_M_count',
    'NM_MUNICIPIO_agrupado_Porto Alegre_sum': 'NM_MUNICIPIO_Porto_Alegre_count',
    'NM_MUNICIPIO_agrupado_Porto Alegre_mean': 'NM_MUNICIPIO_Porto_Alegre_proportion',
    'NM_MUNICIPIO_agrupado_Outros_sum': 'NM_MUNICIPIO_Outros_count',
    'NM_MUNICIPIO_agrupado_Outros_mean': 'NM_MUNICIPIO_Outros_proportion' # This implicitly represents a part of 'interior'
}, inplace=True)

# Display the first few rows and shape of the aggregated DataFrame
print("DataFrame agrupado por ID_CMPT_MOVEL:")
print(df_agrupado.head())
print(f"Shape do DataFrame agrupado: {df_agrupado.shape}")
```
